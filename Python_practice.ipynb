```python
# Constants for the bakery pricing
REGULAR_PRICE = 3.49
DISCOUNT_RATE = 0.60

# 1. Read the number of day-old loaves from the user
loaves_input = input("Enter the number of day-old loaves being purchased: ")
loaves = int(loaves_input)

# 2. Perform financial calculations
regular_total = loaves * REGULAR_PRICE
discount_total = regular_total * DISCOUNT_RATE
total_price = regular_total - discount_total

# 3. Display the formatted receipt with aligned decimals
# :>10.2f spaces out the text and aligns numbers to the right side
print(f"Regular price: ${regular_total:>10.2f}")
print(f"Discount:      ${discount_total:>10.2f}")
print(f"Total price:   ${total_price:>10.2f}")
```

```python
# Define the noise data in descending order
noises = {
    "Jackhammer": 130,
    "Gas Lawnmower": 106,
    "Alarm Clock": 70,
    "Quiet Room": 40
}

# 1. Convert the dictionary into a list of tuples for index-based checking
noise_list = list(noises.items())

# 2. Read the decibel input from the user
user_db = float(input("Enter a decibel level: "))

# 3. Handle Edge Cases (Out of bounds)
if user_db > noise_list[0][1]:
    print(f"That is louder than a {noise_list[0][0]}.")
elif user_db < noise_list[-1][1]:
    print(f"That is quieter than a {noise_list[-1][0]}.")
else:
    # 4. Check for exact matches or intermediate ranges
    matched = False

    # Loop through the list to check exact values
    for name, db in noise_list:
        if user_db == db:
            print(name)
            matched = True
            break

    # If it wasn't an exact match, find which two noises it sits between
    if not matched:
        for index in range(len(noise_list) - 1):
            current_name, current_db = noise_list[index]
            next_name, next_db = noise_list[index + 1]

            # Since the list is descending, check if the input falls between them
            if current_db > user_db > next_db:
                print(f"That is between a {next_name} and a {current_name}.")
                break





```
